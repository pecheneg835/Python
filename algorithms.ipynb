```python
A = {1, 3, 4, 5, 7}
B = {1, 2, 4, 6, 8, 9}

# C-шные функции
print(A | B) # Объеденение
print(A & B) # Пересечение
print(A - B) # Разность

# Собесовский уровень

# Объеденение
print({*A, *B})
# Или
print({x for s in (A, B) for x in s})

# Пересечение
small, large = (A, B) if len(A) < len(B) else (B, A)

print({x for x in small if x in large})

# Разность
print({x for x in A if x not in B})
```

Формат ввода
Первая строка содержит единственное число q - количество запросов.

Далее следует q строк. Каждая из этих строк может иметь один из следующих видов:

Для запроса первого типа - "1 x y" (без кавычек)

Для запросов второго типа - "2 x" (без кавычек)

Для запросов третьего типа - "3 x" (без кавычек)

Формат вывода
Вывод должен состоять из count строк, где count - количество запросов второго типа. Каждая строка должна содержать ответ на соответствующий запрос в формате value (где value - число, которое находится на позиции).

Пример
Ввод

8

1 0 5

2 1

3 1

1 0 6

1 0 7

2 2

1 1 5

2 2

Вывод

5

6

5

```python
list_1 = []
questions = []
q = int(input())
count = 0

for i in range(q):
    questions.append(input().split())

for i in questions:
    if i[0] == "1":
        x = int(i[1])
        y = int(i[2])
        if x != 0:
            list_1.insert(x, y)
        else:
            list_1.insert(0, y)
    if i[0] == "2":
        count += 1
        x = int(i[1])
        print(list_1[x - 1])
    if i[0] == "3":
        x = int(i[1])
        list_1.pop(x - 1)
```

Условие

Дан массив a, состоящий из n чисел. Необходимо найти две пары индексов i₁ < j₁ и i₂ < j₂ таких, что:

значение a[i₁] - a[j₁] минимально возможное;

значение a[i₂] - a[j₂] максимально возможное.

Формат ввода
Первая строка — единственное число n — количество чисел в массиве.

Вторая строка — n разделённых пробелом чисел a[i], где a[i] — число на i-ой позиции в массиве.

Формат вывода
Первая строка — два разделённых пробелом числа i₁ и j₁.

Правила выбора при нескольких подходящих парах:

Выбрать пару с минимальным значением i₁.
Если пар с минимальным i₁ несколько — выбрать пару с минимальным значением j₁.
Вторая строка — два разделённых пробелом числа i₂ и j₂.

Правила выбора при нескольких подходящих парах:

Выбрать пару с минимальным значением i₂.
Если пар с минимальным i₂ несколько — выбрать пару с минимальным значением j₂.

Пример 1

Ввод:

6

2 1 3 5 2 4

Вывод:

2 4

4 5

Пример 2

Ввод:

5

3 2 4 5 6

Вывод:

2 5

1 2

Ограничения:

3 ≤ n ≤ 10⁵

1 ≤ a[i] ≤ 10⁵ для всех 1 ≤ i ≤ n

```python
import numpy as np

n = int(input())
a = np.array([input().split()])

# n = 6
# a = np.array([2, 1, 3, 5, 2, 4])
# 2 4
# 4 5

# n = 5
# a = np.array([3, 2, 4, 5, 6])
# 2 5
# 1 2

mn = float('inf')

for i in range(len(a) - 1):
    num_j = max(a[i + 1:])
    j = (np.where(a[i + 1:] == num_j)[0][0]) + i + 1

    diff = a[i] - num_j
    if diff < mn:
        mn = diff
        i_1 = i
        j_1 = j
    elif diff == mn:
        if i < i_1 or (i == i_1 and j < j_1):
            i_1, j_1 = i, j

mx = float('-inf')

for i in range(len(a) - 1):
    num_j = min(a[i + 1:])
    j = (np.where(a[i + 1:] == num_j)[0][0]) + i + 1

    diff = a[i] - num_j
    if diff > mx:
        mx = diff
        i_2, j_2 = i, j
    elif diff == mx:
        if i < i_2 or (i == i_2 and j < j_2):
            i_2, j_2 = i, j

print(i_1 + 1, j_1 + 1)
print(i_2 + 1, j_2 + 1)
```
