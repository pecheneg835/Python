```python
import numpy as np

def predict(X, w, b):
    return w * X + b

X = np.array([30, 40, 50, 60, 70])
y = np.array([6, 8, 10, 12, 14])

print(predict(X, 0.2, 0))
print(predict(X, 0.1, 3))
print(predict(X, 0.124, 3.0004))
```

```python
def mse(y_true, y_pred):
    return np.mean((y_true - y_pred)**2)

print(mse(y, predict(X, 0.1, 3)))
print(mse(y, predict(X, 0.2, 0)))
print(mse(y, predict(X, 0.124, 3.0004)))
```

```python
def gradient_descent_step(X, y, w, b, learning_rate):
    predictions = predict(X, w, b)
    errors = predictions - y
    dw = 2 * np.mean(X * errors)
    db = 2 * np.mean(errors)
    w_new = w - learning_rate * dw
    b_new = b - learning_rate * db

    return w_new, b_new

print(gradient_descent_step(X, y, 0.1, 3, 0.0001))
```

```python
gradient_descent_steps = int(10e6)
learning_rate = 0.0001
w = 0.1
b = 3
w_new, b_new = gradient_descent_step(X, y, w, b, learning_rate)

for epoch in range(gradient_descent_steps):
    w_new, b_new = gradient_descent_step(X, y, w_new, b_new, learning_rate)

print(w_new, b_new)
y_pred = predict(X, w_new, b_new)
print(y_pred)
print(mse(y, y_pred))
```
